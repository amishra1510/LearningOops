**Question 5**
Create a class Circle. Input the radius through the constructor and calculate the area of the circle. (Use π = 3.14)

```python
class Circle:
    PI = 3.14

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.PI * self.radius ** 2


c = Circle(7)
print(f"Area of Circle: {c.area()}")
```

**Question 6**
Create a class BankAccount. The class should have: Account Number, Account Holder Name, and Balance. Create methods: Deposit, Withdraw, and Display Balance.

```python
class BankAccount:
    def __init__(self, acc_number, holder_name, balance=0):
        self.acc_number = acc_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New Balance: {self.balance}")

    def display_balance(self):
        print(f"Account No: {self.acc_number}, Holder: {self.holder_name}, Balance: {self.balance}")


acc = BankAccount(12345, "Karan Mehta", 1000)
acc.display_balance()
acc.deposit(500)
acc.withdraw(300)
acc.withdraw(5000)
acc.display_balance()
```

**Question 10**
Create a class LibraryBook. Store: Book Name, Author, and Number of Copies. Create methods: Issue Book (decrease copies by 1 if available), Return Book (increase copies by 1), and Display Details.

```python
class LibraryBook:
    def __init__(self, book_name, author, copies):
        self.book_name = book_name
        self.author = author
        self.copies = copies

    def issue_book(self):
        if self.copies > 0:
            self.copies -= 1
            print(f"'{self.book_name}' issued. Copies left: {self.copies}")
        else:
            print(f"'{self.book_name}' is not available.")

    def return_book(self):
        self.copies += 1
        print(f"'{self.book_name}' returned. Copies now: {self.copies}")

    def display_details(self):
        print(f"Book: {self.book_name}, Author: {self.author}, Copies: {self.copies}")


lb = LibraryBook("Wings of Fire", "A. P. J. Abdul Kalam", 3)
lb.display_details()
lb.issue_book()
lb.return_book()
```

**Question 11**
Create a class Calculator with methods: add(), subtract(), multiply(), and divide(). The numbers should be provided through the constructor.

```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 == 0:
            return "Cannot divide by zero"
        return self.num1 / self.num2


calc = Calculator(20, 4)
print(f"Add: {calc.add()}")
print(f"Subtract: {calc.subtract()}")
print(f"Multiply: {calc.multiply()}")
print(f"Divide: {calc.divide()}")
```

**Question 13**
Create a MobilePhone class with attributes brand, model, storage, and battery_percentage. Use a constructor to initialize the details. Implement methods to display phone information, charge the battery, and check the battery status.

```python
class MobilePhone:
    def __init__(self, brand, model, storage, battery_percentage):
        self.brand = brand
        self.model = model
        self.storage = storage
        self.battery_percentage = battery_percentage

    def display_info(self):
        print(f"Brand: {self.brand}, Model: {self.model}, Storage: {self.storage}, "
              f"Battery: {self.battery_percentage}%")

    def charge_battery(self, amount):
        self.battery_percentage = min(100, self.battery_percentage + amount)
        print(f"Battery charged. Current level: {self.battery_percentage}%")

    def check_battery_status(self):
        if self.battery_percentage <= 20:
            print("Battery status: Low, please charge soon.")
        elif self.battery_percentage <= 80:
            print("Battery status: Moderate.")
        else:
            print("Battery status: Good.")


phone = MobilePhone("Samsung", "Galaxy S23", "128GB", 15)
phone.display_info()
phone.check_battery_status()
phone.charge_battery(50)
phone.check_battery_status()
```

**Question 14**
Create a Product class with attributes product_name, product_id, price, and quantity. Implement methods to calculate the total bill and display product details.

```python
class Product:
    def __init__(self, product_name, product_id, price, quantity):
        self.product_name = product_name
        self.product_id = product_id
        self.price = price
        self.quantity = quantity

    def total_bill(self):
        return self.price * self.quantity

    def display_details(self):
        print(f"Product: {self.product_name}, ID: {self.product_id}, "
              f"Price: {self.price}, Quantity: {self.quantity}")
        print(f"Total Bill: {self.total_bill()}")


product = Product("Notebook", "P001", 50, 4)
product.display_details()
```

**Question 15**
Create a BookStore class with attributes book_name, author, price, and number_of_copies. Implement methods to display book details, sell a book, and add new copies.

```python
class BookStore:
    def __init__(self, book_name, author, price, number_of_copies):
        self.book_name = book_name
        self.author = author
        self.price = price
        self.number_of_copies = number_of_copies

    def display_details(self):
        print(f"Book: {self.book_name}, Author: {self.author}, "
              f"Price: {self.price}, Copies: {self.number_of_copies}")

    def sell_book(self, qty=1):
        if self.number_of_copies >= qty:
            self.number_of_copies -= qty
            print(f"Sold {qty} copy/copies. Remaining: {self.number_of_copies}")
        else:
            print("Not enough copies to sell.")

    def add_copies(self, qty):
        self.number_of_copies += qty
        print(f"Added {qty} copies. Total now: {self.number_of_copies}")


store = BookStore("The Alchemist", "Paulo Coelho", 250, 10)
store.display_details()
store.sell_book(3)
store.add_copies(5)
```

**Question 16**
Create an Attendance class with attributes student_name, total_classes, and attended_classes. Implement methods to calculate the attendance percentage and display whether the student is eligible to appear in the examination.

```python
class Attendance:
    def __init__(self, student_name, total_classes, attended_classes):
        self.student_name = student_name
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    def attendance_percentage(self):
        return (self.attended_classes / self.total_classes) * 100

    def display_eligibility(self):
        percentage = self.attendance_percentage()
        print(f"Student: {self.student_name}, Attendance: {percentage:.2f}%")
        if percentage >= 75:
            print("Eligible to appear in the examination.")
        else:
            print("Not eligible to appear in the examination.")


att = Attendance("Simran Kaur", 100, 80)
att.display_eligibility()
```
