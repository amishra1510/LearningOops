Case Study 1: FinTech & Banking (Account Security). Create a BankAccount class with private __balance, public deposit(amount), withdraw(amount), and get_balance(). Attempt to directly overwrite account.__balance and verify via get_balance() that it remained untampered.

```python
class BankAccount:
    def __init__(self, owner, initial_balance):
        self.owner = owner
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. Balance: {self.__balance}")
        else:
            print("Deposit amount must be greater than 0.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. Balance: {self.__balance}")
        else:
            print("Insufficient balance.")

    def get_balance(self):
        return self.__balance

account = BankAccount("Aman", 5000)
account.deposit(2000)
account.withdraw(1000)

account.__balance = 1000000
print("Attempted direct overwrite: account.__balance = 1000000")
print("Actual balance via get_balance():", account.get_balance())
```

Case Study 2: E-Commerce Order Processing (Protected Logistics). Create a parent class Order with protected _shipping_status and a subclass ExpressOrder with expedite_shipping() and track_order() to demonstrate protected members in inheritance.

```python
class Order:
    def __init__(self):
        self._shipping_status = "Processing"

class ExpressOrder(Order):
    def expedite_shipping(self):
        self._shipping_status = "In Transit - Priority"

    def track_order(self):
        return self._shipping_status

order1 = ExpressOrder()
print("Initial Status:", order1.track_order())
order1.expedite_shipping()
print("Updated Status:", order1.track_order())
```

**Question 4**
Case Study 4: Enterprise Payroll System (Preventing Subclass Overwrites). Create a base class BasePayroll with private __tax_rate = 0.20 and calculate_base_tax(salary), and a child class RegionalPayroll that sets __tax_rate = 0.05, demonstrating that name mangling keeps the two variables completely separate.

```python
class BasePayroll:
    def __init__(self):
        self.__tax_rate = 0.20

    def calculate_base_tax(self, salary):
        return salary * self.__tax_rate

class RegionalPayroll(BasePayroll):
    def __init__(self):
        super().__init__()
        self.__tax_rate = 0.05

payroll = RegionalPayroll()
print("Base tax on salary 100000:", payroll.calculate_base_tax(100000))
print("This proves calculate_base_tax() still uses the parent's tax rate (0.20),")
print("because name mangling keeps '_BasePayroll__tax_rate' and '_RegionalPayroll__tax_rate' separate.")
```

Case Study 5: Smart Home IoT Device (Controlled State Changes). Create a Thermostat class with private __target_temp defaulting to 22, set_temperature(temp) validating 16<=temp<=30, and get_temperature(). Demonstrate attempting to set the temperature to 10°C and 24°C.

```python
class Thermostat:
    def __init__(self):
        self.__target_temp = 22

    def set_temperature(self, temp):
        if 16 <= temp <= 30:
            self.__target_temp = temp
            print(f"Temperature set to {temp}°C.")
        else:
            print(f"Warning: {temp}°C is outside the safe range (16-30°C). Temperature not changed.")

    def get_temperature(self):
        return self.__target_temp

thermo = Thermostat()
print("Initial Temperature:", thermo.get_temperature())

thermo.set_temperature(10)
print("Current Temperature:", thermo.get_temperature())

thermo.set_temperature(24)
print("Current Temperature:", thermo.get_temperature())
```

Case Study 7: Digital Wallet Service (Pin-Protected Transactions). Create a DigitalWallet class with private __balance and __pin, implement verify_pin(entered_pin) and transfer_funds(amount, target_wallet, entered_pin), and test both a successful transfer and a failed transfer with an incorrect PIN.

```python
class DigitalWallet:
    def __init__(self, balance, pin):
        self.__balance = balance
        self.__pin = pin

    def verify_pin(self, entered_pin):
        return entered_pin == self.__pin

    def transfer_funds(self, amount, target_wallet, entered_pin):
        if not self.verify_pin(entered_pin):
            print("Transfer failed: Incorrect PIN.")
            return
        if self.__balance < amount:
            print("Transfer failed: Insufficient balance.")
            return
        self.__balance -= amount
        target_wallet.__balance += amount
        print(f"Transferred {amount} successfully.")

wallet1 = DigitalWallet(5000, "1234")
wallet2 = DigitalWallet(1000, "5678")

wallet1.transfer_funds(1000, wallet2, "1234")
wallet1.transfer_funds(1000, wallet2, "0000")
```

Case Study 8: University Grading System (Read-Only Transcripts). Create a StudentTranscript class with public student_id and private __grades dictionary, implement add_grade(subject, grade), calculate_gpa(), and a read-only get_transcript() that returns a copy so external users cannot mutate internal state.

```python
class StudentTranscript:
    def __init__(self, student_id):
        self.student_id = student_id
        self.__grades = {}

    def add_grade(self, subject, grade):
        self.__grades[subject] = grade

    def calculate_gpa(self):
        if not self.__grades:
            return 0
        return sum(self.__grades.values()) / len(self.__grades)

    def get_transcript(self):
        return dict(self.__grades)

t1 = StudentTranscript("102403001")
t1.add_grade("Maths", 9)
t1.add_grade("Physics", 8)
t1.add_grade("OOP", 9)

print("GPA:", t1.calculate_gpa())

transcript = t1.get_transcript()
transcript["Maths"] = 2
print("External copy modified, but internal record is safe:")
print("get_transcript() again:", t1.get_transcript())
```
