**Question 7**
Write a Python program to check whether a number is prime or not.

```python
num = int(input("Enter a number: "))

if num < 2:
    print("Not Prime")
else:
    prime = True

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("Prime")
    else:
        print("Not Prime")
```

**Question 9**
Write a Python program using a function to find the sum of all odd numbers from 1 to n, where n is entered by the user.

```python
def odd_sum(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total

n = int(input("Enter n: "))
print(odd_sum(n))
```

**Question 10**
Write a Python program using a function to find the sum of all prime numbers from 1 to n, where n is entered by the user.

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False

    return True


def prime_sum(n):
    total = 0

    for i in range(2, n + 1):
        if is_prime(i):
            total += i

    return total


n = int(input("Enter n: "))
print(prime_sum(n))
```
