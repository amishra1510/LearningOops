**Question 1**
Consider the Employee class with public name, protected _department, and private __salary. Create an object, modify emp.name and emp._department, and attempt emp.__salary = 70000. Print all three values, then use name mangling to print the original salary. Does emp.__salary = 70000 modify the original private variable?

```python
class Employee:
    def __init__(self):
        self.name = "Ravi"
        self._department = "CSE"
        self.__salary = 50000

emp = Employee()

emp.name = "Amit"
emp._department = "ECE"
emp.__salary = 70000

print("name:", emp.name)
print("_department:", emp._department)
print("__salary:", emp.__salary)
print("Original private salary:", emp._Employee__salary)

print("\nAnswer: No, emp.__salary = 70000 does not modify the original private variable.")
print("It creates a new attribute outside the class because of name mangling.")
```

**Question 4**
Create a DigitalWallet class with public owner_name, protected _wallet_type, and private __balance and __security_code. Implement deposit(amount), withdraw(amount, security_code), check_balance(), and change_security_code(). If the withdrawal amount exceeds the balance, display an appropriate message, and ensure balance/security code cannot be manipulated directly.

```python
class DigitalWallet:
    def __init__(self, owner_name, balance=0, security_code="0000"):
        self.owner_name = owner_name
        self._wallet_type = "Standard"
        self.__balance = balance
        self.__security_code = security_code

    def deposit(self, amount):
        self.__balance += amount
        print(f"Deposited {amount}. New Balance: {self.__balance}")

    def withdraw(self, amount, security_code):
        if security_code != self.__security_code:
            print("Incorrect security code. Withdrawal denied.")
            return
        if amount > self.__balance:
            print("Insufficient balance for withdrawal.")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. New Balance: {self.__balance}")

    def check_balance(self):
        print("Current Balance:", self.__balance)

    def change_security_code(self, old_code, new_code):
        if old_code == self.__security_code:
            self.__security_code = new_code
            print("Security code changed successfully.")
        else:
            print("Old security code incorrect. Cannot change code.")

wallet = DigitalWallet("Aman", 5000, "1234")
wallet.deposit(2000)
wallet.withdraw(10000, "1234")
wallet.withdraw(3000, "1234")
wallet.check_balance()
wallet.change_security_code("1234", "5678")
```

**Question 5**
Design a SmartDoor class with public owner_name, protected _door_status, and private __security_pin. Implement lock_door(), unlock_door(pin), change_pin(old_pin, new_pin), and show_status(). The door starts Locked, unlocks only with the correct PIN, and the PIN can only be changed by providing the old PIN.

```python
class SmartDoor:
    def __init__(self, pin):
        self.owner_name = "Aman"
        self._door_status = "Locked"
        self.__security_pin = pin

    def lock_door(self):
        self._door_status = "Locked"
        print("Door is now Locked.")

    def unlock_door(self, pin):
        if pin == self.__security_pin:
            self._door_status = "Unlocked"
            print("Door Unlocked successfully.")
        else:
            print("Incorrect PIN. Door remains Locked.")

    def change_pin(self, old_pin, new_pin):
        if old_pin == self.__security_pin:
            self.__security_pin = new_pin
            print("PIN changed successfully.")
        else:
            print("Old PIN incorrect. PIN not changed.")

    def show_status(self):
        print("Door Status:", self._door_status)

door = SmartDoor("4321")
door.show_status()
door.unlock_door("0000")
door.unlock_door("4321")
door.show_status()
door.change_pin("4321", "1122")
door.lock_door()
door.show_status()
```

**Question 6**
Case Study 1: E-Commerce Order Processing (Protected Logistics). Create a parent class Order with protected _shipping_status = 'Processing' and a subclass ExpressOrder with expedite_shipping() and track_order() to demonstrate protected members in inheritance.

```python
class Order:
    def __init__(self):
        self._shipping_status = "Processing"

class ExpressOrder(Order):
    def expedite_shipping(self):
        self._shipping_status = "In Transit - Priority"

    def track_order(self):
        return self._shipping_status

order1 = ExpressOrder()
print("Initial Status:", order1.track_order())
order1.expedite_shipping()
print("Updated Status:", order1.track_order())
```

```python
class Taxi:
    def __init__(self, driver, location, contact):
        self.driver = driver
        self._location = location
        self.__contact = contact

    def disp(self):
        print(f"Driver: {self.driver}, Location: {self._location}, Contact: {self.__contact}")

taxi1 = Taxi("Suresh", "MG Road", "9876543210")

try:
    print(taxi1.__contact)
except AttributeError as e:
    print("Direct access error:", e)

taxi1.disp()

print("Admin access via name mangling:", taxi1._Taxi__contact)
```

**Question 8**
Case Study 3: Gaming Engine (Player Health & Invincibility). Create a Player class with public username and private __hp (default 100). Add take_damage(amount), heal(amount), and is_alive(), ensuring hp stays between 0 and 100, and simulate an attempted direct hp manipulation.

```python
class Player:
    def __init__(self, username):
        self.username = username
        self.__hp = 100

    def take_damage(self, amount):
        self.__hp = max(0, self.__hp - amount)
        print(f"{self.username} took {amount} damage. HP: {self.__hp}")

    def heal(self, amount):
        self.__hp = min(100, self.__hp + amount)
        print(f"{self.username} healed {amount} HP. HP: {self.__hp}")

    def is_alive(self):
        return self.__hp > 0

player = Player("Shadow")
player.take_damage(40)
player.heal(10)

player.__hp = 9999
print("Attempted direct HP change: player.__hp = 9999")
print("Is Alive:", player.is_alive())
print("Internal HP is still protected by name mangling.")
```
