**Question 11**
Create a class Student with Name and Marks in four subjects. Create a method to calculate and display Average Marks and Grade (Grade A if average >= 75, else Grade B).

```python
class Student:
    def __init__(self, name, m1, m2, m3, m4):
        self.name = name
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
        self.m4 = m4

    def result(self):
        average = (self.m1 + self.m2 + self.m3 + self.m4) / 4
        grade = "A" if average >= 75 else "B"
        print("Name:", self.name)
        print("Average Marks:", average)
        print("Grade:", grade)

s1 = Student("Amritanshu", 80, 78, 85, 90)
s1.result()
```

**Question 12**
Create a class BankAccount with Account Holder Name, Account Number, and Balance. Create methods Deposit Amount, Withdraw Amount, and Display Balance.

```python
class BankAccount:
    def __init__(self, holder_name, acc_number, balance=0):
        self.holder_name = holder_name
        self.acc_number = acc_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New Balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New Balance: {self.balance}")
        else:
            print("Insufficient funds")

    def display_balance(self):
        print("Account Holder:", self.holder_name)
        print("Account Number:", self.acc_number)
        print("Balance:", self.balance)

acc = BankAccount("Rahul Sharma", "AC1234", 5000)
acc.deposit(2000)
acc.withdraw(3000)
acc.display_balance()
```

**Question 13**
Create an InventoryItem class with attributes item_name, sku, price, and stock_quantity. Implement methods to display item details, purchase item (reduce stock), and restock item (increase stock).

```python
class InventoryItem:
    def __init__(self, item_name, sku, price, stock_quantity):
        self.item_name = item_name
        self.sku = sku
        self.price = price
        self.stock_quantity = stock_quantity

    def display_details(self):
        print(f"Item: {self.item_name}, SKU: {self.sku}, Price: {self.price}, Stock: {self.stock_quantity}")

    def purchase_item(self, quantity):
        if quantity <= self.stock_quantity:
            self.stock_quantity -= quantity
            print(f"Purchased {quantity} units. Remaining stock: {self.stock_quantity}")
        else:
            print("Not enough stock available")

    def restock_item(self, quantity):
        self.stock_quantity += quantity
        print(f"Restocked {quantity} units. Current stock: {self.stock_quantity}")

item = InventoryItem("Wireless Mouse", "SKU001", 799, 50)
item.display_details()
item.purchase_item(10)
item.restock_item(20)
item.display_details()
```

**Question 14**
Create a FitnessTracker class with attributes user_name, daily_step_goal, and steps_walked. Implement methods to calculate progress percentage and display whether the goal was achieved.

```python
class FitnessTracker:
    def __init__(self, user_name, daily_step_goal, steps_walked):
        self.user_name = user_name
        self.daily_step_goal = daily_step_goal
        self.steps_walked = steps_walked

    def progress_percentage(self):
        percentage = (self.steps_walked / self.daily_step_goal) * 100
        print(f"{self.user_name}'s progress: {percentage:.2f}%")
        return percentage

    def goal_status(self):
        if self.steps_walked >= self.daily_step_goal:
            print(f"{self.user_name} has achieved their fitness goal for the day!")
        else:
            print(f"{self.user_name} has not yet achieved their fitness goal for the day.")

tracker = FitnessTracker("Aman", 10000, 8500)
tracker.progress_percentage()
tracker.goal_status()
```

**Question 15**
Case Study: Smart Home Energy Management System. Create a SmartAppliance class with appliance_name, room_location, power_rating_watts, and is_on. Implement turn_on(), turn_off(), and display_status(). Create two objects and confirm their states are managed independently.

```python
class SmartAppliance:
    def __init__(self, appliance_name, room_location, power_rating_watts, is_on=False):
        self.appliance_name = appliance_name
        self.room_location = room_location
        self.power_rating_watts = power_rating_watts
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True
        print(f"{self.appliance_name} in {self.room_location} is now ON.")

    def turn_off(self):
        self.is_on = False
        print(f"{self.appliance_name} in {self.room_location} is now OFF.")

    def display_status(self):
        state = "ON" if self.is_on else "OFF"
        print(f"Appliance: {self.appliance_name}, Room: {self.room_location}, "
              f"Power Rating: {self.power_rating_watts}W, Status: {state}")

ac_unit = SmartAppliance("Air Conditioner", "Living Room", 1500)
smart_light = SmartAppliance("Smart Light", "Bedroom", 15)

ac_unit.turn_on()

ac_unit.display_status()
smart_light.display_status()

print("\nBoth objects manage their 'is_on' state independently.")
```

**Question 16**
Case Study: A smart city Smart Energy Board wants a program to monitor substations using an EnergySubstation class with class variables city_name and total_city_power_drawn, and instance methods add_load() and display_status(), demonstrating overload handling and citywide tracking.

```python
class EnergySubstation:
    city_name = "Neo Metro"
    total_city_power_drawn = 0.0

    def __init__(self, substation_id, max_capacity_kw):
        self.substation_id = substation_id
        self.max_capacity_kw = max_capacity_kw
        self.current_load_kw = 0.0

    def add_load(self, amount):
        if self.current_load_kw + amount <= self.max_capacity_kw:
            self.current_load_kw += amount
            EnergySubstation.total_city_power_drawn += amount
            print(f"{self.substation_id}: Added {amount} kW load. "
                  f"Current load: {self.current_load_kw}/{self.max_capacity_kw} kW.")
        else:
            print(f"{self.substation_id}: Overload warning! Cannot add {amount} kW load.")

    def display_status(self):
        print(f"Substation: {self.substation_id}, "
              f"Load: {self.current_load_kw}/{self.max_capacity_kw} kW, "
              f"City: {EnergySubstation.city_name}")

sub_1 = EnergySubstation("SUB_NORTH", 500.0)
sub_2 = EnergySubstation("SUB_SOUTH", 300.0)

sub_1.add_load(350.0)
sub_2.add_load(200.0)
sub_2.add_load(150.0)

sub_2.max_capacity_kw = 400.0
sub_2.add_load(100.0)

sub_1.display_status()
sub_2.display_status()

print("Total citywide power drawn:", EnergySubstation.total_city_power_drawn)
```
